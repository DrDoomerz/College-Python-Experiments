# Some important theory: Normalization in pd.crosstab

The `normalize` parameter in `pandas.crosstab()` is used to display proportions or percentages instead of raw counts. Here's how its common values work:

*   **`normalize=True` (or `normalize='all'`)**:
    *   Normalizes the entire table by the grand total. Each value in the table represents the proportion of that specific combination relative to *all observations*. The sum of all cells will be 1 (or 100% if multiplied by 100).
    *   **Example:** "What percentage of *all students* are Male and received a Grade 'A'?"

*   **`normalize='index'`**:
    *   Normalizes each row independently by its row sum. Each value represents the proportion relative to its *respective row total*. The sum of values across each row will be 1 (or 100% if multiplied by 100).
    *   **Example:** "*Among all students in the CSE Department*, what percentage received a Grade 'A'?" or "*Among all Female students*, what percentage received a Grade 'B'?"
