```python
# Dataframe 1 Command 2 Part 1
df1.notna()
# This will be the opposite of the boolean value we were getting.
# Here false means that the value is missing.
```

```python
# Importing Dataframe 2
data = {
    "Roll No": [101, 102, 103, 104, 105],
    "Name": ["Asha", "Rohan", "Meera", "Karan", "Anil"],
    "Age": [20, "-", 21, 20, "22"],
    "Marks": [85, 90, "-", 88, 76],
    "Department": ["CSE", "IT", "CSE", "CSE", "E&TC"],
    "Admission_Date": ["01-06-2023", "03-06-2023", "05-06-2023", "07-06-2023", "10-06-2023"]
}

df = pd.DataFrame(data)
df
```

```python
# Dataframe 2 Command 1: Replace '-' with NaN to mark missing values
df.replace("-", np.nan, inplace=True)
df
```

```python
# Dataframe 2 Command 2: Convert blank values to NaN
# This will only work in specific columns. For example, this won't work in the Admission Data column.
df["Age"] = pd.to_numeric(df["Age"], errors="coerce")
df["Marks"] = pd.to_numeric(df["Marks"], errors="coerce")
df
```

```python
# Dataframe 2 Command 3: Checking missing values
print("\nMissing Values:")
print(df.isna().sum())
```

```python
# Dataframe 2 Command 4: Filling the missing age with mean and marks with median
df["Age"].fillna(df["Age"].mean(), inplace=True)
df["Marks"].fillna(df["Marks"].median(), inplace=True)
df
```

```python
# Dataframe 2 Command 5: Fix inconsistent text formatting - converting values to uppercase
df["Department"] = df["Department"].str.upper()
df
```

```python
# Dataframe 2 Command 6: Convert Admission_Date to standard datetime format
df["Admission_Date"] = pd.to_datetime(df["Admission_Date"], errors = "coerce")
df
```
