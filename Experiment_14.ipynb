```python
# Dataset 1 Import
import pandas as pd
import numpy as np

data = {
    'Product': ['Laptop', 'Mobile', 'Tablet', 'Headphones', 'Smartwatch', 'Camera'],
    'Price': [55000, 20000, 30000, 2000, 8000, 45000],
    'Units_Sold': [25, 60, 35, 80, 50, 15]
}

df = pd.DataFrame(data)
print(df)
```

```python
# Data Binning for Dataset 1

bins1 = [0, 10000, 30000, 60000]
labels1 = ['Low', 'Medium', 'High']

df['Price_Category'] = pd.cut(df['Price'], bins=bins1, labels=labels1)

print(df)
```

```python
# Binning Units Sold for Dataset 1
bins2 = [0, 30, 60, 100]
labels2 = ['Low Sales', 'Medium Sales', 'High Sales']

df['Sales_Category'] = pd.cut(df['Units_Sold'], bins=bins2, labels=labels2)

print(df)
```

```python
# Checking Data Types for Dataset 1
print(df.dtypes)
```

```python
# Data Formatting for Dataset 1
# Convert Data Type

print(df.dtypes)

df['Units_Sold'] = df['Units_Sold'].astype(float)

print(df.dtypes)

df
```

```python
# Convert Text to Uppercase for Dataset 1
df['Product'] = df['Product'].str.upper()

print(df)
```

```python
# Rounding Values for Dataset 1
df['Price'] = df['Price'].round(2)

print(df)
```

```python
# Additional Operations for Dataset 1
# Sort data
df.sort_values(by='Price', ascending=False, inplace=True)
print(df)

# Note for self: Python arranges data in ascending order automatically when tasked to.
# Here we are overwriting the table so we used inplace command.
# To make it descending, you cannot use a new command. You have to modify ascending.
```

```python
# Display Unique Values for Dataset 1
print(df['Price_Category'].unique())
```

```python
# Dataset 2 Import: Food Delivery orders
import pandas as pd
import numpy as np

data = {
    'Order_ID': ['01', '02', '03', '04', '05', '06', '07', '08'],
    'Order_Value': [250, 450, 120, 600, 300, 150, 500, 220],
    'Delivery_Time': [30, 45, 25, 60, 35, 20, 50, 28],
    'Distance_km': [5, 8, 3, 10, 6, 2, 9, 4]
}

df = pd.DataFrame(data)

print(df)
```

```python
# Data Binning (Order Value) for Dataset 2
bins_val = [0, 200, 400, 700]
labels_val = ['Low', 'Medium', 'High']

df['Order_Category'] = pd.cut(df['Order_Value'], bins=bins_val, labels=labels_val)
print(df)
```

```python
# Binning Delivery Speed for Dataset 2
bins = [0, 30, 50, 70]
labels = ['Fast', 'Medium', 'Slow']

df['Delivery_Speed'] = pd.cut(df['Delivery_Time'], bins=bins, labels=labels)
print(df)
```

```python
# Binning Delivery Distance for Dataset 2

bins = [0, 3, 7, 12]
labels = ['Short Distance', 'Medium Distance', 'Long Distance']

df['Distance_Category'] = pd.cut(df['Distance_km'], bins=bins, labels=labels)

print(df)
```

```python
# Checking Data Types for Dataset 2
print(df.dtypes)
```

```python
# Data Formatting for Dataset 2
# Convert Data Type

print(df.dtypes)

df['Delivery_Time'] = df['Delivery_Time'].astype(float)

df
```

```python
# Count of Categories for Dataset 2
print("Distance Category Count:\n", df['Distance_Category'].value_counts())
print("\n Delivery Speed Count:\n", df['Delivery_Speed'].value_counts())
print("\n Order Category Count:\n", df['Order_Category'].value_counts())
```

```python
# Sorting By Order Value for Dataset 2
df_sorted = df.sort_values(by='Order_Value')
print(df_sorted)
```

```python
# Convert Text to Uppercase for Order Category in Dataset 2
df['Order_Category'] = df['Order_Category'].str.upper()

print(df)
```
